```python
data['Age']=pd.to_numeric(data['Age'] , errors='coerce')
data['Timestamp']=pd.to_datetime(data['Timestamp'] , format='%d/%m/%Y %H:%M:%S')
```

```python
bins=[0,18,25,35,50,100]
labels=['under 18','18-25','26-35','35-50','50+']
data['Age_Group']=pd.cut(data['Age'], bins=bins, labels=labels)
```

```python
data['Age_Group'].value_counts()
```

```python
col_to_fill=['Gender','Occupation_Status','Country_Region','Shop_Most_Where','Preferred_Platform','Preferred_Payment_Method' ]
```

```python
for col in col_to_fill:
    data[col]=data[col].fillna('Unknown')
```

```python
#
data[col_to_fill].isna().sum()
```

```python
data.columns
```

```python
last6=['Online_Shopping_Satisfaction_Score',
       'Trust_In_Online_Reviews_Score', 'Regret_After_Purchase_Frequency',
       'Reason_Prefer_Online', 'Reason_Prefer_InStore', 'Best_Shopping_Mode',
        ]
```

```python
dropoff=data[last6].isna().all(axis=1)
```

```python
dropoff.sum()
```

```python
data.loc[dropoff,'Timestamp'].dt.date.value_counts().sort_index()
```

```python
pd.crosstab(data['Shop_Most_Where'], data['Reason_Prefer_InStore'] ,dropna=False)
```

```python
age_summary = data.groupby('Age_Group' ,observed=True).agg(
    avg_satisfaction=('Online_Shopping_Satisfaction_Score','mean'),
    avg_trust=('Trust_In_Online_Reviews_Score','mean'),
    avg_regret=('Regret_After_Purchase_Frequency','mean'),
    respodents=('Response_ID','count')
).round(2)

age_summary
```

```python
top_countries=data['Country_Region'].value_counts().head(8).index
top_countries 
```

```python
country_summary = data[data['Country_Region'].isin(top_countries)].groupby(
    'Country_Region', observed=True
).agg(
    avg_satisfaction=('Online_Shopping_Satisfaction_Score', 'mean'),
    bnpl_regular_rate=('Uses_BNPL_Installments', lambda x: (x == 'Yes, regularly').mean()),
    respondents=('Response_ID', 'count')
).round(2)

country_summary
```
